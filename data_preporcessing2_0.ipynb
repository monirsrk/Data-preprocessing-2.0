```python
dataset.drop(columns=['Credit_History'],inplace=True)
```

```python
dataset.columns
```

```python
dataset.info()
```

```python
dataset['Gender'].fillna(dataset['Gender'].mode()[0],inplace=True)
```

```python
dataset['Gender'].isnull().sum()
```

```python
for i in dataset.columns:
  if dataset[i].dtypes==object:
    dataset[i].fillna(dataset[i].mode()[0],inplace=True)
  else:
    dataset[i].fillna(dataset[i].mean(),inplace=True)
```

```python
dataset.isnull().sum()
```

```python
for i in dataset.select_dtypes(include="object").columns:
  dataset[i].fillna(dataset[i].mode()[0],inplace=True)
dataset.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
dataset["encoder gender"]=le.fit_transform(dataset["Gender"])
```

```python
dataset.head(10)
```

```python
dataset["Gender"].value_counts()
```

```python
from sklearn.preprocessing import OneHotEncoder
on=OneHotEncoder()
encoded_array=on.fit_transform(dataset[['Property_Area']]).toarray()
array_df=pd.DataFrame(encoded_array,columns=on.get_feature_names_out(['Property_Area']))
dataset=pd.concat([dataset,array_df],axis=1)
```

```python
dataset.head(10)
```

```python
encoder_array=on.fit_transform(dataset[["Property_Area"]]).toarray()
df_array=pd.DataFrame(encoder_array,columns=on.get_feature_names_out(["Property_Area"]))
dataset=pd.concat([dataset,df_array],axis=1)
dataset.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder
od=OrdinalEncoder()
dataset["encoder_education"]=od.fit_transform(dataset[["Education"]])
dataset.head()
```

```python
sns.boxplot(x=dataset['ApplicantIncome'])
```

```python
sns.distplot(dataset['ApplicantIncome'])
```

```python
q1=dataset["ApplicantIncome"].quantile(0.25)
q3=dataset["ApplicantIncome"].quantile(0.75)
IQR=q3-q1
min_range=q1-1.5*IQR
max_range=q3+1.5*IQR
```

```python
dataset=dataset[dataset["ApplicantIncome"]<=max_range]
sns.boxplot(x=dataset["ApplicantIncome"])
```

```python
from sklearn.preprocessing import StandardScaler
ss=StandardScaler()
dataset["scaler_ApplicantIncome"]=ss.fit_transform(dataset[["ApplicantIncome"]])
dataset.head()
```

```python
from sklearn.preprocessing import MinMaxScaler
ss=MinMaxScaler()
dataset["smmm_ApplicantIncome"]=ss.fit_transform(dataset[["ApplicantIncome"]])
dataset.head()


```
